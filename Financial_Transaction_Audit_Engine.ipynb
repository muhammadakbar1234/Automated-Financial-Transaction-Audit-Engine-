```python
def count_txn_types(df):
    freq_dict = {}
    
    # Loop over each transaction type in the DataFrame column
    for txn in df["Txn_Type"]:
        if txn in freq_dict:
            freq_dict[txn] += 1
        else:
            freq_dict[txn] = 1
            
    return freq_dict

# Test the function on our DataFrame
type_counts = count_txn_types(df)
print(type_counts)
```

```python
def flag_risk(row):
    if pd.notnull(row["Amount_USD"]) and row["Txn_Type"] == "Wire Transfer" and row["Amount_USD"] > 50000:
        return("High Risk")
    else: 
        return("Normal")

# Applying the function across all rows (axis=1 means row-by-row)
df["Risk_Level"] = df.apply(flag_risk, axis=1)

# Displaying the result
print(df[["Txn_ID", "Account_Holders", "Txn_Type", "Amount_USD", "Risk_Level"]])
```

```python
import pandas as pd

class FinancialAuditor:
    def __init__(self, raw_data):
        """Initializes the auditor with raw data"""
        self.df = pd.DataFrame(raw_data)
    def clean_ledger(self):
        """Removes duplicates, fills missing amounts, and parses dates."""
        #1. Drop Duplicates
        self.df = self.df.drop_duplicates()
        # 2. Fill missing amounts with the column mean
        mean_amount = self.df["Amount_USD"].mean()
        self.df["Amount_USD"] = self.df["Amount_USD"].fillna(mean_amount)
        # 3. Convert dates to datetime and extract Day of Week
        self.df["Txn_Date"] = pd.to_datetime(self.df["Txn_Date"])
        self.df["Day_Of_Week"] = self.df["Txn_Date"].dt.day_name()
        return self.df
    def flag_risks(self):
        """Flags high-risk transactions (> $50,000 Wire Transfers)."""
        def check_risk(row):
            if row["Txn_Type"] == "Wire Transfer" and row["Amount_USD"] > 50000:
                return "HIGH_RISK"
            return "NORMAL"   
        self.df["Risk_Level"] = self.df.apply(check_risk, axis=1)
        return self.df
    def get_summary_stats(self):
        """Returns total revenue processed and risk counts."""
        total_volume = self.df["Amount_USD"].sum()
        risk_counts = self.df["Risk_Level"].value_counts().to_dict()
        return {
            "Total_Volume_USD": round(total_volume, 2),
            "Risk_Distribution": risk_counts
        }    


# --- Execution ---
auditor = FinancialAuditor(raw_data)
df_cleaned = auditor.clean_ledger()
df_audited = auditor.flag_risks()
summary = auditor.get_summary_stats()

print(df_audited)
print("\n--- AUDIT SUMMARY ---")
print(summary)
```
